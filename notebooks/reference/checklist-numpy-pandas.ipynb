## Python — база

- [ ] **Списки и словари.** Создание, `len`, обращение по индексу и по ключу,
  перебор в `for`.
- [ ] **Срезы.** `a[1:5]`, `a[:3]`, `a[::-1]` — те же правила потом работают
  в NumPy.
- [ ] **Ветвление и циклы.** `if / elif / else`, `for x in range(n)`,
  `enumerate` (индекс и значение сразу), `zip` (два списка параллельно).
- [ ] **Свои функции.** `def f(x, alpha=1.0): ... return ...` — в заданиях вы
  дописываете именно такие.
- [ ] **f-строки с форматом.** `f"ошибка {err:.3f}"`, `f"{p:.1%}"`. В курсе их
  больше трёхсот: без этого не прочитать почти ни одну ячейку.
- [ ] **Списковые включения.** `[f(x) for x in xs if x > 0]` вместо цикла с
  `append`.
- [ ] **Импорты.** `import numpy as np`, `from sklearn.linear_model import Ridge`.
- [ ] **Чтение ошибок.** `NameError`, `KeyError`, `ValueError`, `IndexError` —
  и умение найти в traceback **свою** строку: она обычно последняя.

## NumPy

- [ ] **Массив и форма.** `np.array([1, 2, 3])`, `X.shape`; `(100, 5)` — это
  100 объектов по 5 признаков.
- [ ] **Срезы по двум осям.** `X[0]` — строка, `X[:, 1]` — столбец,
  `X[2:5, :2]` — кусок.
- [ ] **Булева маска.** `x[x > 0]` — отбор, `(x > 0).sum()` — сколько,
  `(x > 0).mean()` — **доля**.
- [ ] **`axis`.** `X.mean(axis=0)` — по столбцам, `axis=1` — по строкам.
  Самая частая ошибка новичка.
- [ ] **Broadcasting.** `X - X.mean(axis=0)` — вычитание вектора из каждой
  строки без цикла.
- [ ] **Новая ось.** `a[:, None]` превращает вектор `(m,)` в столбец `(m, 1)`.
  Так складывают столбец со строкой и получают матрицу — например, матрицу
  попарных расстояний без единого цикла.
- [ ] **`@` и `.T`.** `X.T @ X` — матричное умножение; `*` умножает поэлементно.
- [ ] **`np.nan`.** `np.nan != np.nan`, поэтому пропуски ищут через `np.isnan`,
  а не сравнением.
- [ ] **Воспроизводимость.** `rng = np.random.default_rng(42)`.
- [ ] **Векторизация вместо цикла.** Если пишете `for i in range(len(X))` —
  почти наверняка есть запись без цикла, короче и в десятки раз быстрее.
- [ ] **Создание массивов.** `linspace`, `arange`, `zeros`, `ones`, `eye`,
  `column_stack`, `reshape`, `ravel`.
- [ ] **Отбор по индексам.** `X[idx]`, где `idx` — массив номеров, а не одно
  число: так берут пакет объектов из выборки. От булевой маски отличается
  тем, что порядок задаёте вы.
- [ ] **Сортировка и поиск.** `argsort`, `argmax`, `where`, `unique`,
  `cumsum`, `diag`.
- [ ] **Линейная алгебра, `np.linalg`.** `solve`, `lstsq`, `norm`, `cond`,
  `eigh`, `svd`. Решение системы пишется как `solve(A, b)`, а не через
  обратную матрицу — на этом стоит половина курса.

## pandas

- [ ] **`DataFrame` и `Series`.** `df.shape`, `df.head()`, `df.dtypes`.
- [ ] **Выбор столбца.** `df["age"]` — `Series`, `df[["age"]]` — `DataFrame`.
- [ ] **`.loc` и `.iloc`.** По имени и по номеру.
- [ ] **Фильтр.** `df[df["age"] > 30]`; несколько условий — через `&`, `|` и
  обязательные скобки.
- [ ] **Пропуски.** `.isna().sum()`, `.dropna()`, `.fillna(значение)`.
- [ ] **Преобразования.** `.astype`, `.map({...})`, `.replace`, `.str.replace`.
- [ ] **Группировка.** `df.groupby("pclass")["survived"].mean()`,
  `.agg(["size", "mean"])`.
- [ ] **Частоты.** `.value_counts()`, `.nunique()`, `pd.crosstab`.
- [ ] **Сводки.** `.describe()`, `.median()`, `.corr()`.
- [ ] **Дубликаты и типы.** `.duplicated().sum()`,
  `.select_dtypes(include="number")`.
- [ ] **Чтение файла.** `pd.read_csv("table.csv")`, а если файл выгружен из
  Excel по-русски — `pd.read_csv(..., sep=";", decimal=",")`.

## Графики

- [ ] **matplotlib.** `fig, ax = plt.subplots()`, `ax.plot`, `ax.scatter`,
  `ax.set_xlabel`, `ax.legend()`, `plt.show()`. Больше не нужно.
- [ ] **seaborn.** Рисует таблицу, а не массивы: `sns.histplot(df, x="age",
  hue="survived")`. Ключевой приём — `hue`: одним словом превращает график в
  сравнение групп.

## scikit-learn

- [ ] **Единый интерфейс.** У всех моделей и преобразований он один: `fit`
  учит, `predict` предсказывает, `transform` преобразует, `score` считает
  качество. Выучив его на одной модели, вы умеете пользоваться всеми.
- [ ] **Честное разбиение.** `train_test_split`, `Pipeline` — чтобы обработка
  данных не подглядывала в контрольную часть.
