### Задание 1.2. Почему на практике всегда SVD

Собственные числа $C$ равны квадратам сингулярных чисел $X$ (делённым на $\ell$),
поэтому $\mathrm{cond}(C) = \mathrm{cond}(X)^2$ — построив $C$ явно, мы
**возводим обусловленность в квадрат** и теряем вдвое больше верных знаков. Ровно
тот же эффект, что с нормальными уравнениями в домашней работе 2.

**Важно:** матрицу надо собирать **уже центрированной**, иначе центрирование
внутри `fit` изменит спектр и сравнивать будет не с чем.

### Задание 1.3. Почему `lstsq` выживал: МНК через псевдообратную

В домашней работе 2 вы видели, что на матрице Вандермонда нормальные уравнения
разваливаются при $\mathrm{cond}(X)\approx10^8$, а `lstsq` держится. Тогда мы
не могли объяснить почему — нужного аппарата ещё не было. Теперь он есть.

Замечание 8.13: подставив $X = U\Sigma W^{\mathsf T}$ в решение теоремы 1.24,
получаем $X^{\mathsf T}X = W\Sigma^2W^{\mathsf T}$ и

$$
\theta^* = (X^{\mathsf T}X)^{-1}X^{\mathsf T}y
= W\Sigma^{-2}W^{\mathsf T}\,W\Sigma U^{\mathsf T}y
= W\Sigma^{-1}U^{\mathsf T}y = X^{+}y,
$$

где $X^{+} = W\Sigma^{-1}U^{\mathsf T}$ — **псевдообратная** матрица. Разница
не в результате, а в пути: справа $X^{\mathsf T}X$ не строится вовсе, поэтому
обусловленность не возводится в квадрат. Именно это `lstsq` и делает внутри.
